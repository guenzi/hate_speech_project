```python
import ast

df = pd.read_csv("../data/final_datasets/hatexplain.csv")

RACIAL_TARGETS = {"African", "Asian", "Jewish", "Arab", "Caucasian", "Hispanic", "Indian", "Islam"}

# Parser les targets
df["targets"] = df["targets"].apply(ast.literal_eval)

# Garder mono-ethnie
df = df[df["targets"].apply(
    lambda t: len(set(t) & RACIAL_TARGETS) == 1
)].reset_index(drop=True)

# Extraire l'ethnie unique
df["ethnicity"] = df["targets"].apply(
    lambda t: list(set(t) & RACIAL_TARGETS)[0]
)

# Distribution par ethnie et label
print(df.groupby(["ethnicity", "label"]).size().unstack(fill_value=0))
```
